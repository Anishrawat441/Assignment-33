# Task 1: Create  SQLite Database
## 1. Create a SQLite Database file named company db.
## 2. Create the following tabels:
## Table 1: employees id (INTEGER,PRIMARY KEY)
## name (TEXT) department(TEXT)
## salary (INTEGER)
## Table 2: sales 
## sale_id(INTEGER,PRIMARY KEY)
## employee id (INTEGER)
## amount(INTEGER)
## sale_date(TEXT)

```python

import sqlite3

# 1. Create SQLite database
connection = sqlite3.connect("company.db")

# Create cursor
cursor = connection.cursor()

# 2. Create employees table
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary INTEGER
)
""")

# 3. Create sales table
cursor.execute("""
CREATE TABLE IF NOT EXISTS sales (
    sale_id INTEGER PRIMARY KEY,
    employee_id INTEGER,
    amount INTEGER,
    sale_date TEXT
)
""")

# Save changes
connection.commit()

print("Database and tables created successfully!")

# Close connection
connection.close()
```

```python

from sqlalchemy import create_engine, inspect

# 1. Create SQLAlchemy engine
engine = create_engine("sqlite:///company.db")

# 2. Test database connection
with engine.connect() as connection:
    print("Database connected successfully!")

# 3. Fetch table names
inspector = inspect(engine)
tables = inspector.get_table_names()

print("Tables in company.db:")

for table in tables:
    print(table)

# 4. Close the engine
engine.dispose()
```

```python
from langchain_community.utilities import SQLDatabase
from langchain_community.agent_toolkits import SQLDatabaseToolkit
from langchain_groq import ChatGroq
import os
```

```python
db = SQLDatabase.from_uri("sqlite:///company.db")

print(db.dialect)
print(db.get_usable_table_names())
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
)
```

```python
toolkit = SQLDatabaseToolkit(
    db=db,
    llm=llm
)

tools = toolkit.get_tools()

print("Available SQL Tools:")
for tool in tools:
    print(tool.name)
    print(tool.description)
```

```python
## Explore built in tools
query_tool = next(
    tool for tool in tools
    if tool.name == "sql_db_query"
)

result = query_tool.invoke(
    "SELECT * FROM employee LIMIT 5;"
)

print(result)
```

```python
## Get table information
schema_tool = next(
    tool for tool in tools
    if tool.name == "sql_db_schema"
)

result = schema_tool.invoke("sqlite:///company.db")

print(result)
```

# Task 6: Create SQL Agent
## 1. Create an SQL Agent using langchain.
## 2. Enable verbose mode to observe:
## . Thought process
## . Generated SQL queries
## 3. Ensure the agent can choose tools autometically.
## 4. This agent should be able to communicate with  both SQLite and Workbench Data.


```python
from langchain_community.utilities import SQLDatabase
from langchain_community.agent_toolkits import create_sql_agent
from langchain_groq import ChatGroq
```

```python
import os
from dotenv import load_dotenv

load_dotenv()

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
)
```

```python
sqlite_db = SQLDatabase.from_uri(
    "sqlite:///company.db"
)

print(sqlite_db.get_usable_table_names())
```

```python
sqlite_agent = create_sql_agent(
    llm=llm,
    db=sqlite_db,
    verbose=True,
    agent_type="tool-calling",
    allow_dangerous_requests=True
)
```

```python
response = sqlite_agent.invoke({
    "input": "Show all employees from the database"
})

print(response["output"])
```

```python
## Connect mysql workbench
from langchain_community.utilities import SQLDatabase
from sqlalchemy import URL

mysql_url = URL.create(
    drivername="mysql+pymysql",
    username="root",
    password="Ani123",
    host="localhost",
    port=3306,
    database="company_db"
)

mysql_db = SQLDatabase.from_uri(mysql_url)

print(mysql_db.get_usable_table_names())
```

```python
mysql_agent = create_sql_agent(
    llm=llm,
    db=mysql_db,
    verbose=True,
    agent_type="tool-calling",
    allow_dangerous_requests=True
)
```

```python
response = mysql_agent.invoke({
    "input": "Show all tables in the database"
})

print(response["output"])
```

```python
response = mysql_agent.invoke({
    "input": "Find the top 5 students based on marks"
})

print(response["output"])
```

```python
## USING both with one sql agent
from langchain_community.agent_toolkits.sql.toolkit import (
    SQLDatabaseToolkit
)

# SQLite toolkit
sqlite_toolkit = SQLDatabaseToolkit(
    db=sqlite_db,
    llm=llm
)

# MySQL toolkit
mysql_toolkit = SQLDatabaseToolkit(
    db=mysql_db,
    llm=llm
)

# Combine tools
tools = (
    sqlite_toolkit.get_tools()
    + mysql_toolkit.get_tools()
)

# Create one agent
from langchain.agents import create_agent

sql_agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt="""
    You are an SQL database assistant.

    You have access to two databases:
    1. SQLite database
    2. MySQL Workbench database

    Select the appropriate database tools based
    on the user's question.

    SQLite tools are for the local SQLite database.
    MySQL tools are for the MySQL database.

    Always inspect the database schema before
    generating SQL queries.

    Never execute destructive queries such as
    DROP, DELETE or UPDATE without explicit approval.

    Provide clear and accurate answers.
    """
)
```

```python
response = sql_agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "Show all tables in SQLite database"
        }
    ]
})

print(response["messages"][-1].content)
```
